```python
from pathlib import Path

MODEL_PATH = Path(
    r"D:\Karthick babu\Learning\9.Deep learning\Capstone project-v1\runs\vehicle_detection\combined_army_vs_nonarmy\weights\best.pt"
)

print("Model exists:", MODEL_PATH.exists())
```

```python
from ultralytics import YOLO
from pathlib import Path
import matplotlib.pyplot as plt
import cv2

# Load your combined model
model = YOLO(str(MODEL_PATH))
print("Classes:", model.names)

# Enter the full path of an image
IMAGE_PATH = input("Enter vehicle image path: ").strip().strip('"')

if not Path(IMAGE_PATH).is_file():
    print("Image not found. Check the path.")
else:
    print("Detecting vehicle...")

    results = model.predict(
        source=IMAGE_PATH,
        conf=0.5,
        imgsz=320,
        device="cpu",
        verbose=False
    )

    result = results[0]
    non_army_found = False

    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])
        label = model.names[class_id]

        print(f"{label}: {confidence:.1%}")

        if class_id == 1:
            non_army_found = True

    if non_army_found:
        print("WARNING: Non-Army Vehicle Detected!")
    else:
        print("No Non-Army Vehicle detected above 50% confidence.")

    # Display the annotated image in Jupyter
    annotated_image = result.plot()
    annotated_image = cv2.cvtColor(
        annotated_image, cv2.COLOR_BGR2RGB
    )

    plt.figure(figsize=(12, 7))
    plt.imshow(annotated_image)
    plt.axis("off")
    plt.show()

    # Save the output
    output_path = Path(IMAGE_PATH).with_name(
        "vehicle_detection_output.jpg"
    )
    cv2.imwrite(str(output_path), result.plot())
    print("Output saved to:", output_path)
```

```python

import cv2
from ultralytics import YOLO

# 1. Load your trained YOLO model
model = YOLO(str(MODEL_PATH))

# 2. Start laptop webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Webcam not found!")
else:
    print("Show your vehicle picture to the webcam.")
    print("Press Q to stop.")

    while True:
        ret, frame = cap.read()

        if not ret:
            break

        # 3. Detect vehicles in the webcam frame
        results = model.predict(
            source=frame,
            conf=0.25,
            imgsz=320,
            device="cpu",
            verbose=False
        )

        result = results[0]
        non_army_found = False

        # 4. Read predictions
        for box in result.boxes:
            class_id = int(box.cls[0])
            confidence = float(box.conf[0])
            label = model.names[class_id]

            if class_id == 1:
                non_army_found = True

        # 5. Draw prediction boxes
        output_frame = result.plot()

        # 6. Display warning
        if non_army_found:
            cv2.putText(
                output_frame,
                "WARNING: NON-ARMY VEHICLE!",
                (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 0, 255),
                2
            )
        else:
            cv2.putText(
                output_frame,
                "Monitoring Vehicle...",
                (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

        cv2.imshow("AI Vehicle Early-Warning System", output_frame)

        # Press Q to exit
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    cap.release()
    cv2.destroyAllWindows()
```
